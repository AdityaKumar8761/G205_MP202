```python
numeric_cols = [
    "age_years",
    "km_driven",
    "engine_cc",
    "owners",
    "service_records",
    "accidents"
]
for col in numeric_cols:
    print(f"\n========== {col} ==========")
    print(
        df.groupby("city_zone")[col]
        .agg(["count", "min", "mean", "median", "max"])
        .round(2)
    )
```

```python
df["date"] = pd.to_datetime(df["date"])
numeric_cols = [
    "age_years", "km_driven", "engine_cc",
    "owners", "service_records", "accidents"
]
daily = df.groupby("date")
for col in numeric_cols:
    stats = daily[col].agg(["count", "nunique"])
    suspicious = stats[
        (stats["count"] >= 3) &
        (stats["nunique"] == 1)
    ]
    print(f"\n{col}: {len(suspicious)} suspicious dates")
    if not suspicious.empty:
        print(suspicious.head(20))
```

```python
suspicious = df[
    (df["date"] >= "2026-04-10") &
    (df["date"] <= "2026-05-02")
]
print("Accident values during suspicious period:")
print(
    suspicious.groupby("date")["accidents"]
    .agg(["count", "min", "max", "nunique"])
    .to_string()
)
```

```python
print("Target Statistics:")
print(df["resale_thousand_rs"].describe())
print("\nHighest 20 resale prices:")
display(
    df[
        ["record_id", "resale_thousand_rs"]
    ]
    .sort_values(
        "resale_thousand_rs",
        ascending=False
    )
    .head(20)
)
```

```python
import pandas as pd
import numpy as np
clean_df = df.copy()
print("Before Cleaning:", clean_df.shape)
```

```python
clean_df["service_records"] = clean_df["service_records"].replace(-999, np.nan)
print("Missing service records:")
print(clean_df["service_records"].isnull().sum())
```

```python
# Ensure date column is datetime
clean_df["date"] = pd.to_datetime(clean_df["date"])
mask = clean_df["date"].between(
    "2026-04-11",
    "2026-04-29"
)
print("Affected rows:", mask.sum())
clean_df.loc[mask, "accidents"] = np.nan
```

```python
# Standardize text
clean_df["brand_tier"] = (
    clean_df["brand_tier"]
    .str.strip()
    .str.lower()
    .str.replace(".", "", regex=False)
)
clean_df["brand_tier"] = clean_df["brand_tier"].replace({
    "bud": "budget",
    "pre": "premium"
})
print(clean_df["brand_tier"].value_counts())
```

```python
print(clean_df.isnull().sum())
```

```python
clean_df = clean_df.drop(
    columns=["final_price_from_rc_transfer"]
)
```

```python

from pathlib import Path
import pandas as pd
import numpy as np

def clean_data(df):
    df = df.copy()

    # 1. Replace sentinel values
    df["service_records"] = df[
        "service_records"
    ].replace(-999, np.nan)

    # 2. Handle stuck sensor
    df["date"] = pd.to_datetime(df["date"])

    mask = df["date"].between(
        "2026-04-11", "2026-04-29"
    )
    df.loc[mask, "accidents"] = np.nan

    # 3. Standardize brand names
    df["brand_tier"] = (
        df["brand_tier"]
        .str.strip()
        .str.lower()
        .str.replace(".", "", regex=False)
        .replace({
            "bud": "budget",
            "pre": "premium"
        })
    )

    # 4. Remove leakage column
    df = df.drop(
        columns=["final_price_from_rc_transfer"],
        errors="ignore"
    )

    return df


if __name__ == "__main__":
    raw = pd.read_csv("train.csv")
    cleaned = clean_data(raw)

    output = Path("data/processed/cleaned_train.csv")
    output.parent.mkdir(parents=True, exist_ok=True)

    cleaned.to_csv(output, index=False)

    print("Cleaning completed!")
    print("Shape:", cleaned.shape)
    print("Missing values:")
    print(cleaned.isnull().sum())
```
